## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target / proxy: trend_direction

I will use trend_direction as a simple proxy for identifying pages that may need attention. Pages with a downward trend are more relevant to the refresh decision.

The target can be represented as a binary label:
- 1 = declining page
- 0 = not declining

This is only a practical proxy for prioritization and does not mean that a declining trend was caused by content freshness.

A higher Precision@K means that more of the pages ranked near the top are relevant to the content review decision.

One row represents one web page. Each page has observable search and content signals such as impressions, average position, CTR, content age, and word count. The model uses these page-level signals to rank pages for content-refresh review.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule can identify pages using a few simple conditions, such as high impressions and old content. However, it may miss combinations of several signals. ML can learn patterns from multiple features such as content age, impressions, average position, CTR, and word count, and use them to rank pages by refresh priority. This can make the prioritization more flexible than a single hand-written rule.
